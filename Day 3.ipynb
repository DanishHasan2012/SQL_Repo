### Day 3

You are analyzing a social network dataset at Google. Your task is to find mutual friends between two users, Karl and Hans. There is only one user named Karl and one named Hans in the dataset.

The output should contain 'user_id' and 'user_name' columns.

```sql
USE CATALOG sql;
```

```sql
%skip
CREATE TABLE sql.day3.users(user_id INT, user_name varchar(30));
INSERT INTO sql.day3.users VALUES (1, 'Karl'), (2, 'Hans'), (3, 'Emma'), (4, 'Emma'), (5, 'Mike'), (6, 'Lucas'), (7, 'Sarah'), (8, 'Lucas'), (9, 'Anna'), (10, 'John');

CREATE TABLE sql.day3.friends(user_id INT, friend_id INT);
INSERT INTO sql.day3.friends VALUES (1,3),(1,5),(2,3),(2,4),(3,1),(3,2),(3,6),(4,7),(5,8),(6,9),(7,10),(8,6),(9,10),(10,7),(10,9);
```

```sql
--solution:
with karl_frnds as (
  select friend_id from friends
  where user_id = (select user_id from users where user_name = 'Karl')
),
hans_frnds as (
  select friend_id from friends
  where user_id = (select user_id from users where user_name = 'Hans')
)
select distinct f.friend_id, u.user_name as friend_name
from friends f join users u on f.friend_id = u.user_id
join karl_frnds kf on u.user_id = kf.friend_id
join hans_frnds hf on u.user_id = hf.friend_id;
```

```sql
-- We need to do double join
-- 1. u.user_id = f.user_id --> To get the user's user name
-- 2. f.friend_id = uf.user_id --> To get the Friend's user name.

select u.user_id as user_id, u.user_name as user_name,
f.friend_id as friend_id, uf.user_name as friend_name from users u 
join friends f on u.user_id = f.user_id
join users uf on f.friend_id = uf.user_id
order by u.user_id, f.friend_id;
```
