Output the nationality along with the corresponding total number of available beds. Sort records by the total available beds in descending order.

```sql
%sql
--Find the total number of available beds per hosts' nationality.
--logic : join both tables. group by nationality, count(n_beds).

select ah.nationality, sum(n_beds) as total_n_beds from sql.day6.airbnb_apartments aa join sql.day6.airbnb_hosts ah on aa.host_id = ah.host_id 
group by ah.nationality order by sum(n_beds) desc;
```

```sql
%python
from pyspark.sql import functions as F
```

```sql
%python
airbnb_hosts = spark.read.table("sql.day6.airbnb_hosts")
airbnb_apartments = spark.read.table("sql.day6.airbnb_apartments")
df_join = airbnb_apartments.join(airbnb_hosts, on = "host_id", how = "inner")
df_result = df_join.groupBy("nationality").agg(F.sum("n_beds").alias("total_n_beds")).orderBy(F.desc("total_n_beds")).show()
```
