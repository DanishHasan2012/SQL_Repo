```sql
--calculate by the amount of project budget allocated to each employee.
select p.title, p.budget/count(emp.emp_id) OVER(partition by p.id) as budget_per_employee
from day5.ms_projects p join day5.ms_emp_projects emp on p.id = emp.project_id
order by budget_per_employee desc;
```
